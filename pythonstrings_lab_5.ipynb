#string methods
#1.split
#2.join
#3.strip
#4.upper and lower
#5.replace
#6.find
#7.count

```python
text="hello world,welcome to python "

words=text.split()
print("1. split():",words)

joined_text="-".join(words)
print("2.join():",joined_text)

stripped_text=text.strip()
print("3. strip():",f"'{stripped_text}'")

print("4a. upper():",stripped_text.upper())
print("4b.lower():",stripped_text.upper().lower())

replaced_text=stripped_text.replace("python","coding")
print("5. replace():",replaced_text)

position=stripped_text.find("world")
print("6.find():",position)

character_count=stripped_text.count("o")
print("7.count():",character_count)
```

#program that count no of vowels and consonants in a string

```python
a= str(input())
b=a.lower()
vowels="aieou"
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
     count_v+=1
    else:
      count_c+=1
print("the count of vowels is ",count_v)
print("the count of consonants is",count_c)
```
