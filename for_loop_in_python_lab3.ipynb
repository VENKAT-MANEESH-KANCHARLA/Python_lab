#print prime numbers between the 2no's

```python
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end=" ")
```

#print the sum of the digits of a given number

```python
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```
