#arithematic operators

```python
x=15
y=5
print(x+y)
print(x-y)
print(x*y)
print(x/y)
print(x%y)
print(x**y)
print(x//y)
```

#assignment operators

```python
a=10
print(a)
a+=5
print(a)
a-=5
print(a)
a*=5
print(a)
a/=5
print(a)
```

#ternary operators

```python
n=int(input())
result="positive" if n>0 else "negative"
print(result)
```

```python
# comparision operators
x=10
y=20
print(x==y)
print(x!=y)
print(x>y)
print(x<y)
print(x>=y)
print(x<=y)
```

```python
# logical operators
x=5
print(x>5 and x<10)
print(x>5 or x<10)
print(not(x>5 and x<10))
```

```python
# identity operator
a=10
b=a
c=20
print(a is b)
print(a is c)
print(a is not c)
```

```python
# membership operator
a=[1,2,3,4,5,6]
print(2 in a)
print(4 in a)
print(8 in a)
```

```python
# bitwise operator
a=10
b=20
print(a&b)
print(a|b)
print(a^b)
print(~a)
print(a<<1)
print(a>>1)
```
