#program to find greatest among three


```python
a=input("enter a value")
b=input("enter b value")
c=input("enter c value")
if(a>b and a>c):
  print(f"{a} is greater")
elif(b>c):
  print(f"{b} is greater")
else:
  print(f"{c} is greater")
```

#program on match case

```python
day=int(input("enter any number"))
match day:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wednesday")
  case 4:
    print("thursday")
  case 5:
    print("friday")
  case 6:
    print("saturday")
  case 7:
    print("sunday")
  case _:
    print("default")
```

#program that print "python" 5 times
