#List operations


```python

data = ["Python", "Java", "C++", "Rust"]

print("1. Indexing:", data[0])


print("2. Slicing:", data[1:3])


print("3. Concatenation:", data + ["Go", "Swift"])

print("4. Repetition:", ["Code"] * 3)


print("5. Membership ('Java' in data):", "Java" in data)

print("6. Iteration:")
for item in data:
    print("  -", item)
```

#functions on lists

```python
numbers = [15, 3, 42, 8, 23]

# 1. Built-in aggregation functions: len(), min(), max(), sum()
print("Original List:", numbers)
print("1. len():", len(numbers))
print("   min():", min(numbers))
print("   max():", max(numbers))
print("   sum():", sum(numbers))

# 2. Adding elements: append(), insert(), extend()
numbers.append(100)
numbers.insert(1, 50)
numbers.extend([60, 70])
print("\n2. After append, insert, extend:", numbers)

# 3. Searching and counting: index(), count()
print("\n3. Index of 42:", numbers.index(42))
print("   Count of 50:", numbers.count(50))

# 4. Removing elements: remove(), pop()
numbers.remove(15)
popped_item = numbers.pop()
print("\n4. After remove(15) and pop():", numbers)
print("   Popped item:", popped_item)

# 5. Reordering: sort(), reverse()
numbers.sort()
print("\n5. Sorted List:", numbers)
numbers.reverse()
print("   Reversed List:", numbers)

# 6. Clearing list: clear()
numbers.clear()
print("\n6. After clear():", numbers)
```
