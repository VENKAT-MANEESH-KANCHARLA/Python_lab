```python
def rev_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
print(rev_str("maneesh"))
```

```python
def multiple_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5,6)
print(f"sum is{res_sum},count is {res_count},the avg is{res_avg}")
```
